```python
df["nvidia_gpu"] = (
    df["gpu_clean"]
    .str.replace(r"Geforce", "GeForce", case=False, regex=True)
    .str.extract(
        r"((?:GeForce\s+)?(?:RTX|GTX|MX)\s*\d+(?:\s*Ti)?(?:\s*Max-Q)?)",
        expand=False
    )
)
```

```python
df["nvidia_gpu"] = (
    df["nvidia_gpu"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
df.loc[
    df["gpu_clean"].str.contains("NVIDIA", case=False, na=False),
    ["graphics", "gpu_clean", "nvidia_gpu"]
].head(30)
```

No rows are returned. Therefore, all NVIDIA configurations in the dataset have a successfully extracted NVIDIA model.

```python
df["gpu_model"] = df["nvidia_gpu"]
```

##### AMD

```python
amd_mask = (
    df["gpu_brand"].eq("AMD")
    & df["gpu_model"].isna()
)

df.loc[amd_mask, "gpu_model"] = (
    df.loc[amd_mask, "gpu_clean"]
    .str.extract(
        r"(Radeon\s+(?:RX\s+)?[A-Za-z0-9]+)",
        expand=False
    )
)
```

##### Intel

```python
intel_mask = (
    df["gpu_brand"].eq("Intel")
    & df["gpu_model"].isna()
)

df.loc[intel_mask, "gpu_model"] = (
    df.loc[intel_mask, "gpu_clean"]
    .str.extract(
        r"(Arc(?:\s+Graphics)?(?:\s+\d+[A-Z])?|"
        r"Iris Xe Graphics|"
        r"Iris Plus Graphics|"
        r"UHD Graphics(?:\s+\d+)?|"
        r"Graphics)",
        expand=False
    )
)
```

##### Apple

```python
apple_mask = (
    df["gpu_brand"].eq("Apple")
    & df["gpu_model"].isna()
)

df.loc[apple_mask, "gpu_model"] = (
    df.loc[apple_mask, "gpu_clean"]
    .str.extract(
        r"(M[1-5](?:\s+(?:Pro|Max))?|A18 Pro)",
        expand=False
    )
)
```

##### Qualcomm

```python
qualcomm_mask = (
    df["gpu_brand"].eq("Qualcomm")
    & df["gpu_model"].isna()
)

df.loc[qualcomm_mask, "gpu_model"] = "Adreno"
```

##### ARM

```python
arm_mask = (
    df["gpu_brand"].eq("ARM")
    & df["gpu_model"].isna()
)

df.loc[arm_mask, "gpu_model"] = (
    df.loc[arm_mask, "gpu_clean"]
    .str.replace("ARM ", "", regex=False)
)
```

Finally, I will remove the redundant Graphics suffix from the extracted model names.

```python
df["gpu_model"] = (
    df["gpu_model"]
    .str.replace(r"\s*Graphics$", "", regex=True)
    .str.strip()
    .replace("", pd.NA)
)
```

##### GPU manufacturer indicators

In addition to gpu_brand, I will create binary indicators showing whether the graphics configuration contains NVIDIA, AMD, or Intel graphics.

These features are useful because a laptop may contain more than one GPU. For example, a configuration such as NVIDIA GeForce RTX 5060 + Intel UHD Graphics belongs to NVIDIA and Intel simultaneously.

```python
df["has_nvidia_gpu"] = df["gpu_clean"].str.contains(
    "NVIDIA",
    case=False,
    na=False
)

df["has_amd_gpu"] = df["gpu_clean"].str.contains(
    "AMD",
    case=False,
    na=False
)

df["has_intel_gpu"] = df["gpu_clean"].str.contains(
    "Intel",
    case=False,
    na=False
)
```

For NVIDIA configurations, the specific GPU model was extracted into `nvidia_gpu`. A unified `gpu_model` feature was then created for the different GPU manufacturers. When the source text did not contain a specific model, the value was kept missing rather than assigning an unsupported model. For example, Intel Graphics does not provide enough information to determine a specific GPU model.

Finally, binary indicators `has_nvidia_gpu`, `has_amd_gpu`, and `has_intel_gpu` were created to identify the presence of each manufacturer independently. This is particularly useful for configurations containing multiple GPUs.
